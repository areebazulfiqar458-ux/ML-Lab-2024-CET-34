Task No 10

```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed -= 5
        
    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket" 
    else:
        return "Big Ticket"

print("Q10: ")
print(caught_speeding (60,False))
print(caught_speeding(75,True))
print(caught_speeding(85,False))
```
